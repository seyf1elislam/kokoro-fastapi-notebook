```python
#@title # Preparing (Clone repo and  Install dependencies , setting .env )
!apt-get update && apt-get install -y espeak-ng libsndfile1

# Clone repository
# !git clone https://github.com/remsky/Kokoro-FastAPI
!git clone https://github.com/seyf1elislam/Kokoro-FastAPI
%cd Kokoro-FastAPI
!git clone https://huggingface.co/hexgrad/Kokoro-82M

!pip install fastapi uvicorn onnxruntime
!pip install -r requirements.txt
!pip install --force-reinstall tensorflow


# writing .env
%cd /content/Kokoro-FastAPI
content="""MODEL_DIR=./Kokoro-82M
PUBLIC_API=True
"""
with open(".env","w") as f:
  f.write(content)
```
